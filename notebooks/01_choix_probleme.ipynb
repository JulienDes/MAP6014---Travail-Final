# Étape 1 — Choix du problème

Ce notebook présente notre choix de problème pour le travail du cours MAP6014, à la session d'automne 2026. Il explique notre motivation, la pertinence du sujet et l'accès aux données. Cette étape demande de choisir un problème de science des données sur Kaggle, avec des données libres d'accès et téléchargeables sans restriction particulière.

## Notre motivation

Nous avons choisi un problème lié au prix des maisons, car nous sommes deux jeunes qui souhaitent éventuellement acheter une propriété. Ce sujet nous intéresse personnellement et nous permet de relier un projet concret à notre apprentissage. Nous souhaitons ainsi mieux comprendre les caractéristiques associées au prix de vente d'une maison, tout en mettant en pratique les méthodes de régression.

## Le problème choisi sur Kaggle

Nous avons retenu la compétition [House Prices – Advanced Regression Techniques](https://www.kaggle.com/competitions/house-prices-advanced-regression-techniques/overview) sur Kaggle. Le problème consiste à prédire le prix de vente d'une maison à partir de ses caractéristiques. Il s'agit d'un problème de régression, puisque la valeur à prédire, `SalePrice`, est un montant numérique.

## L'accès aux données

Les fichiers nécessaires au projet ont été téléchargés depuis Kaggle et placés dans le dossier `data/`. Nous disposons des données d'entraînement (`train.csv`), des données de test (`test.csv`), d'une description des variables (`data_description.txt`) et d'un exemple de fichier de soumission (`sample_submission.csv`).

## La pertinence pour le cours

Ce problème permet de réaliser les principales étapes d'un projet de science des données : explorer les données, préparer les variables, entraîner des modèles de régression et évaluer leurs prédictions. Nous pourrons également comparer différentes approches de prétraitement et de modélisation, puis justifier nos choix à partir des résultats obtenus.

## Les limites du sujet

Les données concernent des propriétés situées à Ames, aux États-Unis. Les observations du projet ne pourront donc pas être directement généralisées au marché immobilier de notre région. Notre objectif est d'étudier les caractéristiques associées aux prix et de produire des prédictions.
